In section 4.3 we discuss **surprisal**. Although surprisal is a term from information theory, we will examine it from the perspective of variational inference in this notebook. In Bayes' theorem we have:

$$
    p(x \mid y) = \frac{p(x, y)}{\sum_x p(x, y)} = \frac{p(x, y)}{p(y)}
$$

We can also rewrite $p(y)$ as an expected value: $p(y) = \mathbb{E}_{p(x)} \left [p(y \mid x) \right ]$. $p(y)$ is the probability of the data itself. In other words, based on the agent's model, $p(y)$ tells us the distribution of data it expects to receive from the environment on average across any possible state it could be in. $p(y)$ goes by many names including **marginal likelihood** and **evidence**. 

Suprisal is the negative log of the evidence, in other words, $-\log p(y)$. One thing that is not obvious due to the notation is that surprisal is calculated in reference to an outcome of a random variable. In other words, we would ask the surprisal of a particular data point, $-\log p(y = \hat{y})$. We can, of course, compute the surprisal for any possible outcome of a probability distribution.

Since surprisal is a negative of a log probability its interpretation is the inverse of probability. Specifically, a low probability event has high surprisal and a high probability event has low surprisal. This relationship can be easily seen by examining a plot of surprisal and evidence side by side. First, we define a function that returns a negative log of a probability for a series of values.

```python
def surprisal(py):
    return -np.loge(py)
```
